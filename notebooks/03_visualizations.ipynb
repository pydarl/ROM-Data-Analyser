```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
df = pd.read_csv('../data/processed/df_clean')
df.head()
```

```python
a = df["knee_angle_deg"]
b = df["Rehabilitation_Time_weeks"]
```

```python
plt.plot(b,a)
```

```python
plt.plot(a, b, 'o')
```

```python
plt.scatter(a, b)
```

```python
c = df.loc[0:5,"Weight_kg"]
d = df.loc[0:5, "Rehabilitation_Time_weeks"]
plt.plot(c,d)
plt.xlabel("Weight(kg)")
plt.ylabel("Rehabilitation Time (Weeks)")
plt.savefig("../figures/weight_vs_rehab_time.png", dpi=300, bbox_inches='tight')
```

```python
plt.hist(a)
plt.xlabel("Knee Angle (Degrees)")
plt.ylabel("Frequency")
plt.savefig("../figures/knee_angle_histogram.png", dpi=300, bbox_inches='tight')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Top-left
axes[0, 0].plot(df["Weight_kg"], df["Rehabilitation_Time_weeks"])
axes[0, 0].set_title("Weight vs Rehab Time")
axes[0, 0].set_xlabel("Weight (kg)")
axes[0, 0].set_ylabel("Rehab Time (weeks)")

# Top-right
axes[0, 1].hist(df["Weight_kg"], bins=10, color='steelblue')
axes[0, 1].set_title("Weight Distribution")
axes[0, 1].set_xlabel("Weight (kg)")
axes[0, 1].set_ylabel("Frequency")

# Bottom-left
axes[1, 0].scatter(df["Age"], df["Rehabilitation_Time_weeks"], color='orange')
axes[1, 0].set_title("Age vs Rehab Time")
axes[1, 0].set_xlabel("Age")
axes[1, 0].set_ylabel("Rehab Time (weeks)")

# Bottom-right
axes[1, 1].bar(df["Position"].value_counts().index, df["Position"].value_counts().values, color='green')
axes[1, 1].set_title("Position Count")
axes[1, 1].set_xlabel("Position")
axes[1, 1].set_ylabel("Count")

plt.tight_layout()  # prevents overlapping labels
plt.savefig("../figures/overview_grid.png", dpi=300, bbox_inches='tight')
```
