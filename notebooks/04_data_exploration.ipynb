```python
def improve_grade(row):
    if row.percent_improvement >= 80:
        return 'Excellent'
    elif row.percent_improvement >= 60:
        return 'Good'
    elif row.percent_improvement >= 40:
        return 'Fair'
    else:        return 'Poor'

df.improvement_grade = df.apply(improve_grade, axis=1)
df["improvement_grade"].value_counts()
```
